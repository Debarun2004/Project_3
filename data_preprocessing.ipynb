```python
import requests
import folium
```

```python
# Example coordinates
node_a = (88.3639, 22.5726) # (lon, lat)
node_b = (88.4021, 22.5855)

# 1. Ask OSRM for the real driving route
osrm_url = f"http://router.project-osrm.org/route/v1/driving/{node_a[0]},{node_a[1]};{node_b[0]},{node_b[1]}?overview=full&geometries=geojson"
response = requests.get(osrm_url).json()

# 2. Extract the curvy real-world geometry (Polyline)
route_geometry = response['routes'][0]['geometry']['coordinates']
route_lat_lon = [[coord[1], coord[0]] for coord in route_geometry]

# 3. Create the map using Esri World Street Map (No API key required, allows local files)
m = folium.Map(location=[22.579, 88.383], zoom_start=13, tiles=None)

folium.TileLayer(
    tiles="https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}",
    attr="Tiles © Esri — Source: Esri"
).add_to(m)

# 4. Add your route on top
folium.PolyLine(route_lat_lon, color="blue", weight=5, opacity=0.8).add_to(m)

# Save to view the perfectly traced road
m.save("traced_route.html")
```
